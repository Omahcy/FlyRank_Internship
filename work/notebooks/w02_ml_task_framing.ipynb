## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
import numpy as np

# Unit of Analysis: 1 row = 1 published content post
data = {
    'post_id': ['p_001', 'p_002', 'p_003', 'p_004', 'p_005'],
    'topic': ['health', 'tech', 'lifestyle', 'health', 'education'],
    'hook_word_count': [14, 8, 22, 11, 7],
    'publish_hour': [9, 14, 18, 8, 20],
    'views': [10500, 3200, 15400, 890, 4200],
    'likes': [820, 110, 1450, 45, 230],
    'shares': [130, 12, 310, 4, 25],
    'comments': [95, 8, 140, 2, 18]
}

df = pd.DataFrame(data)

# Calculate engagement efficiency score (likes + 2*shares + 1.5*comments per view)
df['engagement_rate'] = (df['likes'] + (df['shares'] * 2.0) + (df['comments'] * 1.5)) / df['views']

# Define target: 1 if in top 20% (high performing), else 0
threshold = df['engagement_rate'].quantile(0.80)
df['is_high_performer'] = (df['engagement_rate'] >= threshold).astype(int)

print("Unit of analysis: 1 row = 1 published content post")
print(f"Dataset shape: {df.shape}")
df[['post_id', 'topic', 'hook_word_count', 'views', 'engagement_rate', 'is_high_performer']]
```
